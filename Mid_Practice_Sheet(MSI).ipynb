```python
data= input("Sir please enter your data: ")
alphabets=0
digits=0
special_characters=0
for i in data:
    if  i in "0123456789":
        digits+=1
    elif i ==" " or i == "." :
        special_characters+=1
    else:
        alphabets+=1
print(f"The number of Alphabets in the string is: {alphabets}")
print(f"The number of Digits in the string is : {digits}")
print(f"The number of Special characters in the string is: {special_characters}")
```

```python
# It is a string with the smallest and largest word.

data="It is a string with the smallest and largest word"#input("Sir please enter your data: ")
store=""
count=0
large=""
small=""
for i in data:
    if i not in" .":
        store+=i
        count+=ord(i)
    else:
        if large=="" and small=="":
            small=store
            large=store
            store=""
            count=0
        else:
            #--------------------for large number --------
            temp=0
            for j in large:
                temp+=ord(j)
            if count>temp:
                large=store
           #---------------For small number---------------
            tempu=0
            for k in small:
                tempu+=ord(k)
            if count<tempu:
                small=store
        #---------------------- Reseting store and count
            store=""
            count=0

print(f"The largest word is “{large}” and the smallest word is '{small}'.")



```
