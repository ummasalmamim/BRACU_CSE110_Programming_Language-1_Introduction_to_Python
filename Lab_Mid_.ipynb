```python
x=(input("Sir,please enter your data: "))           #Input--> $$hello$$,##{WORLD2023}#
store=""
for i in x :
    if ord(i)<=90 and ord(i)>=65:
        store+=chr(ord(i)+32)
    elif ord(i)<=122 and ord(i)>=97:
        store+=i
    elif i in "0123456789":
        store+="!"
    elif i==",":
        print(store)
        store=""
print(store)

```
