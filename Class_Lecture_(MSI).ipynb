**List has 2 properties.\
1.Heterogeneous\
2.Mutability**



**Heterogeneous\
(It means list can contain multiple data type.Such as string,float,integer,none type, boolean,and so on.)**

```python
a_list=["Einstein",40,"Audi A8",44.44,None,True]
print(type(a_list))
print(type(a_list[0]))
print(type(a_list[1]))
print(type(a_list[2]))
print(type(a_list[3]))
print(type(a_list[4]))
print(type(a_list[5]))
```

**Mutability\
(It means the value inside the list can be changed.That is,we can change any values in list. So,list mutable.)**



```python
a_list=["Einstein",40,"Audi A8",44.44]                    #  positive indexing
a_list[0]="Newton"
a_list[2]="Tesla"
print(a_list)
```

```python
a_list=["Einstein",40,"Audi A8",44.44]                     #  negative indexing
a_list[-4]="Newton"
a_list[-2]="Tesla"
print(a_list)
```

**Way to access one by one value in list\
Indexing**


```python
a_list=["Einstein",40,"Audi A8",44.44]    # if we want to see money then we can get access money by using index value
print(a_list[3])                          #positve index
```

By using **index** we can access any type of values from anywhere

```python
a_list=["Einstein",40,"Audi A8",44.44]       # if we want to see car and age then we can get access car,age by using index value
print(a_list[-2])
print(a_list[-3])                           # negative index
```

**Way to access values in list\
For Loop**

```python
a_list=["Einstein",40,"Audi A8",44.44]      #if we want to see all vales together,then we can get access values by using for loop
for i in a_list:
    print(i)
```

```python
a_list=["Einstein",40,"Audi A8",44.44]
for i in a_list:
    print(i,end=",")
```

**Sharing Same Memory Location**

```python
a_list=["Einstein",40,"Audi A8",44.44]
dublicate_list= a_list
print(a_list)
print(dublicate_list)
print("==================================")

a_list[0]="Newton"
print(a_list)
print(dublicate_list)
print("==================================")

dublicate_list[2]="Tesla"
print(a_list)
print(dublicate_list)
```

**Membership Operator**

```python
a_list=[20,40,100,1000]
print(40 in a_list)
print(40 not in a_list)
```

```python
a_list=[20,40,100,1000]
print(10 in a_list)
print(10 not in a_list)
```
