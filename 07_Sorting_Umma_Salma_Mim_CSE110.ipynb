```python
#Task-9

import math
points =[(1,7), (4,5), (-1,7), (-2,0), (1,1), (5,-1)]

temp=points.copy()
temporarydict={}

for i in temp:
    point_a,point_b=i
    calculation=math.sqrt((point_a**2)+(point_b**2))
    temporarydict.update({i:calculation})

final_temp=sorted(temporarydict.values())
final={}
for j in final_temp:
    for k in temporarydict.keys():
        if temporarydict[k]==j:
            final[k]=temporarydict[k]

keys_final=[]
values_final=[]

for l in final.keys():
    keys_final.append(l)

for m in final.values():
    values_final.append(m)

print("Minimum distance =",values_final[0])
print("Here the closest point is",keys_final[0],"which has a distance of", values_final[0],"from the origin")
```
